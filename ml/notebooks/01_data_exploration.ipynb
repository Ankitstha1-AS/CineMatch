# CineMatch — MovieLens exploratory data analysis

Phase 1 deliverable. The reusable implementation lives in `ml/eda.py`
(`python -m ml.eda <dataset-dir>`); this notebook is the presentation layer for
the report and produces the same figures under `docs/eda/<dataset>/`.

Run it from the repository root with the ML virtual environment selected as the
kernel. Raw MovieLens files are gitignored and are never committed.

```python
from pathlib import Path

from IPython.display import Markdown, display

from ml.eda import build_bundle, render_summary

ROOT = Path.cwd()
if not (ROOT / "ml").is_dir():
    ROOT = ROOT.parents[1]

DATASET = ROOT / "data" / "processed" / "ml-25m-subset"
OUT = ROOT / "docs" / "eda" / DATASET.name
OUT.mkdir(parents=True, exist_ok=True)

bundle = build_bundle(DATASET)
summary = bundle["summary"]
display(Markdown(render_summary(summary, [])))
```

```python
from IPython.display import Image

from ml.eda import plot_figures

for name in plot_figures(bundle, OUT):
    display(Markdown(f"**{name}**"))
    display(Image(filename=str(OUT / name)))
```

## Findings

- The full 25M release is extremely concentrated: the top 1% of rated movies hold
  47.6% of ratings and the popularity Gini is 0.942.
- The fixed, seeded subset (1,478,848 ratings; 9,960 users; 8,737 movies) keeps the
  long tail while bringing Gini down to 0.712, close to ml-latest-small.
- The subset is 98.3% sparse, so evaluation must rank against all unrated items.
- Only 2.6% of retained users have 20 or fewer ratings, so the cold-start experiment
  must simulate sparse users by revealing 1–50 ratings rather than relying on the
  sample.
- Tags cover 60.9% of subset movies; genres cover the full catalogue. ml-1m is the
  genres-only benchmark.
